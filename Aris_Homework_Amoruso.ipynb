```python
import operator
from operator import itemgetter
def person_lister(f):
    def inner(people):
        people.sort(key=lambda x: int(x[2]))
        return [f(elem) for elem in people]
    return inner

@person_lister
def name_format(person):
    return ("Mr. " if person[3] == "M" else "Ms. ") + person[0] + " " + person[1]

if __name__ == '__main__':
    people = [input().split() for i in range(int(input()))]
    print(*name_format(people), sep='\n')
```

```python
def wrapper(f):
    def fun(l):
        m = []
        for elem in l:
            if len(elem) == 13 and elem[:3] == "+91":
                numero_base = elem[3:]
            elif len(elem) == 12 and elem[:2] == "91":
                numero_base = elem[2:]
            elif len(elem) == 11 and elem[:1] == "0":
                numero_base = elem[1:]
            elif len(elem) == 10:
                numero_base = elem
            formattato = f"+91 {numero_base[:5]} {numero_base[5:]}"
            m.append(formattato)
        return f(m)
    return fun

@wrapper
def sort_phone(l):
    print(*sorted(l), sep='\n')

if __name__ == '__main__':
    l = [input() for _ in range(int(input()))]
    sort_phone(l)
```

```python
import xml.etree.ElementTree as etree

maxdepth = 0
def depth(elem, level):
    global maxdepth
    level+=1
    if level>maxdepth:
        maxdepth=level
    for child in elem:
        depth(child,level)



if __name__ == '__main__':
    n = int(input())
    xml = ""
    for i in range(n):
        xml =  xml + input() + "\n"
    tree = etree.ElementTree(etree.fromstring(xml))
    depth(tree.getroot(), -1)
    print(maxdepth)
```

```python
n = int(input())
def change(s):
  risultato = []
  i = 0
  lunghezza = len(s)

  while i < lunghezza:
    if (
        s[i : i + 2] == "&&"
        and (i > 0 and s[i - 1] == " ")
        and (i + 2 < lunghezza and s[i + 2] == " ")
    ):
      risultato.append("and")
      i += 2
    elif (
        s[i : i + 2] == "||"
        and (i > 0 and s[i - 1] == " ")
        and (i + 2 < lunghezza and s[i + 2] == " ")
    ):
      risultato.append("or")
      i += 2
    else:
      risultato.append(s[i])
      i += 1

  print("".join(risultato))

for _ in range(n):
  change(input())
```

```python
#!/bin/python3

import math
import os
import random
import re
import sys

# Complete the time_delta function below.
from datetime import datetime
def time_delta(t1, t2):
    # Formato corrispondente a: "Sun 10 May 2015 13:54:36 -0700"
    fmt = "%a %d %b %Y %H:%M:%S %z"

    dt1 = datetime.strptime(t1, fmt)
    dt2 = datetime.strptime(t2, fmt)

    # La differenza tra datetime timezone-aware tiene conto del fuso orario
    delta_secondi = int(abs((dt1 - dt2).total_seconds()))
    return str(delta_secondi)

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    t = int(input())

    for t_itr in range(t):
        t1 = input()

        t2 = input()

        delta = time_delta(t1, t2)

        fptr.write(delta + '\n')

    fptr.close()
```

```python
def average(array):
    l=list(array)
    e=set()
    for elem in l:
        e.add(elem)
    x=len(e)
    somma=0
    for elem in e:
        somma=somma+elem
    average=somma/x
    average=round(average,3)
    return average

if __name__ == '__main__':
    n = int(input())
    arr = list(map(int, input().split()))
    result = average(arr)
    print(result)
```

```python
def minion_game(string):
    voc=["A","E","I","O","U"]
    punt_k=0
    punt_s=0
    indice_k=0
    indice_s=0
    for c in string:
        if c in voc:
            punt_k=punt_k+(len(string)-indice_k)
        else:
            punt_s=punt_s+(len(string)-indice_s)
        indice_k+=1
        indice_s+=1
    if punt_k>punt_s:
        print("Kevin",punt_k)
    elif punt_s>punt_k:
        print("Stuart",punt_s)
    else:
        print("Draw")
if __name__ == '__main__':
    s = input()
    minion_game(s)
```

```python
def print_rangoli(size):
    beto=["a","b","c","d","e","f","g","h","i","j","k","l","m","n","o","p","q","r","s","t","u","v","w","x","y","z"]
    lettera=beto[:size]
    lettera=lettera[::-1]
    width=((size*2)-1)+((size*2)-2)
    salvate=[]
    for i in range(size):
        new=lettera[i]
        if i>0:
            for j in range(1,i+1):
                new=lettera[i-j]+"-"+new+"-"+lettera[i-j]
        salvate.append(new)
        linea=new.center(width,"-")
        print(linea)
    salvate.pop()
    for i in range(1,size):
        elem=salvate[-i]
        linea=elem.center(width,"-")
        print(linea)


if __name__ == '__main__':
    n = int(input())
    print_rangoli(n)
```

```python
def count_substring(string, sub_string):
    indice=0
    y=0
    for car in string:
        if car==sub_string[0] and len(sub_string)<=len(string[indice:]):
            x=car
            for j in range(1,len(sub_string)):
                x=x+string[indice+j]
            if x==sub_string:
                y+=1
        indice+=1
    return y

if __name__ == '__main__':
    string = input().strip()
    sub_string = input().strip()

    count = count_substring(string, sub_string)
    print(count)
```

```python
def mutate_string(string, position, character):
    f=list(string)
    f[position]=character
    stringa=""
    for i in f:
        stringa=stringa+i
    return stringa

if __name__ == '__main__':
    s = input()
    i, c = input().split()
    s_new = mutate_string(s, int(i), c)
    print(s_new)
```

```python
def swap_case(s):
    stringa=""
    for car in s:
        if car.islower():
            stringa=stringa+car.upper()
        elif car.isupper():
            stringa=stringa+car.lower()
        else:
            stringa=stringa+car
    return stringa

if __name__ == '__main__':
    s = input()
    result = swap_case(s)
    print(result)
```
